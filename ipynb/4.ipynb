## 前回の復習

実数のリストを受け取り，各要素の絶対値を並べた新しいリストを返す関数を実行する．

```python
def absolute_values(numbers):
    results = []

    for number in numbers:
        if number >= 0:
            results.append(number)
        else:
            results.append(-number)

    return results

numbers = [-3.5, 2.0, -1.2, 0.0, 4.8]
values = absolute_values(numbers)
print(values)
```

## 前回の課題の解答例

**課題1：得点を評価へ分類する関数**

```python
def classify_score(score):
    if not (0 <= score <= 100):
        return "範囲外"

    if score >= 95:
        return "S"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

scores = [-1, 0, 59, 60, 69, 70, 79, 80, 94, 95, 100, 101]

for score in scores:
    grade = classify_score(score)
    print(f"{score}点：{grade}")
```

**課題2：合格者数を数える関数**

```python
def count_passes(scores):
    count = 0

    for score in scores:
        if score >= 60:
            count = count + 1

    return count

scores = [45, 60, 72, 58, 90, 81]
passed = count_passes(scores)
print(f"合格者は{passed}人です．")
```

## ファイル操作を関数にまとめる

ファイル名と文字列のリストを引数として受け取り，各文字列を1行ずつ書き込む．

```python
def write_lines(filename, lines):
    with open(filename, "w", encoding="utf-8") as file:
        for line in lines:
            file.write(f"{line}\n")

topics = ["条件分岐", "繰り返し", "関数", "ファイル操作"]
write_lines("topics.txt", topics)
```

## 課題1

文字列のリストをテキストファイルへ保存する関数 `save_plan(filename, plans)` を作成せよ．

1. 仮引数filenameとplansを持つ関数を定義せよ．
2. 関数の中でfilenameが表すファイルを書き込みモードで開き，UTF-8を指定せよ．
3. for文でplansから文字列を一つずつ取り出し，改行文字を付けて1項目につき1行で書き込め．
4. `plans = ["授業の復習をする", "課題を期限までに提出する", "毎週本を読む"]` を用意せよ．
5. `save_plan("plan.txt", plans)` を呼び出せ．
6. plan.txtを読み込み，保存した3行を表示せよ．

**課題1の解答例**

```python
def save_plan(filename, plans):
    with open(filename, "w", encoding="utf-8") as file:
        for plan in plans:
            file.write(f"{plan}\n")

plans = ["授業の復習をする", "課題を期限までに提出する", "毎週本を読む"]
save_plan("plan.txt", plans)

with open("plan.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

## 課題2

CSVファイルに保存された最高気温の平均を返す関数 `average_temperature(filename)` を作成せよ．

1. 月曜18.5度，火曜20.0度，水曜21.5度，木曜19.0度，金曜23.5度というデータをweek_temperature.csvへ書き込め．
2. 仮引数filenameを持つ関数を定義せよ．
3. 関数の中でcsv.readerを使ってCSVファイルを読み込み，next関数でヘッダーを読み飛ばせ．
4. for文で各行を取り出し，最高気温をfloat関数で実数へ変換せよ．
5. 最高気温の合計とデータ数を求め，平均を戻り値として返せ．
6. 関数を呼び出し，「平均最高気温は20.5度です．」と表示せよ．

**課題2の解答例**

```python
import csv

rows = [
    ["曜日", "最高気温"],
    ["月", 18.5],
    ["火", 20.0],
    ["水", 21.5],
    ["木", 19.0],
    ["金", 23.5],
]

with open("week_temperature.csv", "w", encoding="utf-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

def average_temperature(filename):
    total = 0.0
    count = 0

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        next(reader)

        for row in reader:
            temperature = float(row[1])
            total = total + temperature
            count = count + 1

    return total / count

average = average_temperature("week_temperature.csv")
print(f"平均最高気温は{average}度です．")
```
