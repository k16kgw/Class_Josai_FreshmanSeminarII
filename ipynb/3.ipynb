## 演習1　引数と戻り値を使わない関数

```python
def show_message():
    print("関数を呼び出しました．")

show_message()
show_message()
```

## 演習2　引数を受け取る関数

```python
def show_square(number):
    print(number ** 2)

show_square(3)
show_square(5)
```

## 演習3　mathモジュールと戻り値

```python
import math

def circle_area(radius):
    area = math.pi * radius ** 2
    return area

print(circle_area(3))
print(circle_area(5))
```

## 演習4　実数のリストから絶対値のリストを作る関数

```python
def absolute_values(numbers):
    results = []

    for number in numbers:
        if number >= 0:
            results.append(number)
        else:
            results.append(-number)

    return results

numbers = [-3.5, 2.0, -1.2, 0.0, 4.8]
values = absolute_values(numbers)
print(values)
```

## 課題1　得点を評価へ分類する関数

```python
def classify_score(score):
    if not (0 <= score <= 100):
        return "範囲外"

    if score >= 95:
        return "S"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"

    return "F"

scores = [-1, 0, 59, 60, 69, 70, 79, 80, 94, 95, 100, 101]

for score in scores:
    grade = classify_score(score)
    print(f"{score}点：{grade}")
```

## 課題2　合格者数を数える関数

```python
def count_passes(scores):
    count = 0

    for score in scores:
        if score >= 60:
            count = count + 1

    return count

scores = [45, 60, 72, 58, 90, 81]
passed = count_passes(scores)
print(f"合格者は{passed}人です．")
```
