```python
import random

random.seed(10)
trials = 10000
inside = 0

for _ in range(trials):
    x = random.random()
    y = random.random()

    if x ** 2 + y ** 2 <= 1:
        inside += 1

pi = 4 * inside / trials
print(pi)
```

```python
import math
import random

def approximate_pi(trials):
    inside = 0
    for _ in range(trials):
        x = random.random()
        y = random.random()
        if x ** 2 + y ** 2 <= 1:
            inside += 1
    return 4 * inside / trials

random.seed(10)
for trials in [1000, 10000, 100000]:
    value = approximate_pi(trials)
    print(trials, value, abs(value - math.pi))
```

```python
import random

def monte_carlo_integral(f, a, b, trials):
    total = 0
    for _ in range(trials):
        x = random.uniform(a, b)
        total += f(x)
    return (b - a) * total / trials
```

```python
import random

def f(x):
    return x ** 2

random.seed(10)
for trials in [1000, 10000, 100000]:
    value = monte_carlo_integral(f, 0, 1, trials)
    print(trials, value, abs(value - 1 / 3))
```

```python
import random

def f(x):
    return x ** 2

for seed in [10, 20, 30]:
    random.seed(seed)
    value = monte_carlo_integral(f, 0, 1, 10000)
    print(seed, value, abs(value - 1 / 3))
```
