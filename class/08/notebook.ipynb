```python
import random

random.seed(10)
trials = 6000
counts = [0, 0, 0, 0, 0, 0]

for _ in range(trials):
    value = random.randint(1, 6)
    counts[value - 1] += 1

print(counts)
```

```python
import random

random.seed(10)
trials = 1000
counts = [0, 0, 0, 0, 0, 0]

for _ in range(trials):
    value = random.randint(1, 6)
    counts[value - 1] += 1

for face in range(1, 7):
    print(f"{face}: {counts[face - 1]}回，相対度数 {counts[face - 1] / trials}")
```

```python
import random

random.seed(10)
trials = 10000
inside = 0

for _ in range(trials):
    x = random.random()
    y = random.random()

    if x ** 2 + y ** 2 <= 1:
        inside += 1

pi = 4 * inside / trials
print(pi)
```
