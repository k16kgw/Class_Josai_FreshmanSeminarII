```python
def f(x):
    return x ** 2 - 2

def grid_root(n):
    best_x = 1
    best_residual = abs(f(best_x))

    for i in range(n + 1):
        x = 1 + i / n
        residual = abs(f(x))

        if residual < best_residual:
            best_x = x
            best_residual = residual

    return best_x

exact = 2 ** 0.5

for n in [1, 10, 100, 1000]:
    value = grid_root(n)
    print(n, value, abs(value - exact))
```

```python
def series_sum(n):
    total = 0

    for k in range(1, n + 1):
        total = total + 1 / 2 ** k

    return total

for n in [1, 2, 5, 10]:
    value = series_sum(n)
    print(n, value, abs(value - 1))
```
