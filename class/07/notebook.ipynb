```python
def f(x):
    return x ** 2 - 2

def grid_root(n):
    best_x = 1
    best_residual = abs(f(best_x))

    for i in range(n + 1):
        x = 1 + i / n
        residual = abs(f(x))

        if residual < best_residual:
            best_x = x
            best_residual = residual

    return best_x

exact = 2 ** 0.5

for n in [1, 10, 100, 1000]:
    value = grid_root(n)
    print(n, value, abs(value - exact))
```

```python
def series_sum(n):
    total = 0

    for k in range(1, n + 1):
        total = total + 1 / 2 ** k

    return total

for n in [1, 2, 5, 10]:
    value = series_sum(n)
    print(n, value, abs(value - 1))
```

```python
def forward_difference(f, x, h):
    return (f(x + h) - f(x)) / h

def f(x):
    return x ** 2

value = forward_difference(f, 3, 0.001)
print(value)
```

```python
def central_difference(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)

value = central_difference(f, 3, 0.001)
print(value)
```

```python
def f(x):
    return x ** 2

h = 1e-3
forward = forward_difference(f, 3, h)
center = central_difference(f, 3, h)
print(forward, abs(forward - 6))
print(center, abs(center - 6))
```

```python
import math

def f(x):
    return math.sin(x)

x = math.pi / 4
exact = math.cos(x)

for h in [1e-1, 1e-2, 1e-3, 1e-4]:
    value = central_difference(f, x, h)
    print(h, value, abs(value - exact))
```

```python
def riemann_sum(f, a, b, n):
    width = (b - a) / n
    area = 0

    for i in range(n):
        x = a + i * width
        area = area + f(x) * width

    return area
```

```python
def f(x):
    return x ** 2

value = riemann_sum(f, 0, 1, 100)
print(value)
```

```python
def trapezoidal(f, a, b, n):
    h = (b - a) / n
    total = (f(a) + f(b)) / 2

    for i in range(1, n):
        x = a + i * h
        total = total + f(x)

    return h * total
```

```python
value = trapezoidal(f, 0, 1, 100)
print(value)
```

```python
def f(x):
    return x ** 2

exact = 1 / 3
riemann_value = riemann_sum(f, 0, 1, 100)
trapezoidal_value = trapezoidal(f, 0, 1, 100)
print(riemann_value, abs(riemann_value - exact))
print(trapezoidal_value, abs(trapezoidal_value - exact))
```

```python
for n in [10, 100, 1000, 10000]:
    error_r = abs(riemann_sum(f, 0, 1, n) - exact)
    error_t = abs(trapezoidal(f, 0, 1, n) - exact)
    print(n, error_r, error_t)
```
